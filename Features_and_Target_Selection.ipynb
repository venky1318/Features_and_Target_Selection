```python
# Task 31: Features and Target Selection

## Objective
Select suitable input features and a target variable
from a student performance dataset.

## Deliverables
- Selected input features (X)
- Selected target variable (y)
- Explanation of feature selection
- Training and testing data split
```

```python
import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split

print("Libraries imported successfully!")
```

```python
# Student Performance Dataset

data = {
    "student_id": [
        f"S{i:03d}" for i in range(1, 21)
    ],

    "department": [
        "CSE", "ECE", "IT", "CSE", "ECE",
        "IT", "CSE", "ECE", "IT", "CSE",
        "ECE", "IT", "CSE", "ECE", "IT",
        "CSE", "ECE", "IT", "CSE", "ECE"
    ],

    "study_hours": [
        1.5, 2.0, 2.5, 3.0, 3.5,
        4.0, 4.5, 5.0, 5.5, 6.0,
        6.5, 7.0, 7.5, 8.0, 2.5,
        4.5, 6.0, 7.0, 3.0, 5.5
    ],

    "attendance_pct": [
        65, 70, 72, 75, 78,
        80, 82, 85, 88, 90,
        92, 95, 96, 98, 74,
        84, 89, 93, 77, 87
    ],

    "assignment_score": [
        50, 55, 58, 62, 65,
        68, 70, 74, 78, 82,
        85, 88, 91, 95, 60,
        73, 80, 87, 64, 79
    ],

    "previous_exam_score": [
        45, 50, 52, 58, 60,
        65, 68, 70, 75, 78,
        82, 85, 88, 92, 55,
        68, 76, 83, 59, 74
    ],

    "final_exam_score": [
        48, 53, 56, 60, 64,
        67, 70, 75, 78, 82,
        85, 89, 92, 96, 58,
        72, 79, 87, 62, 80
    ]
}

# Create DataFrame
df = pd.DataFrame(data)

print("Student performance dataset created successfully!")

print("Total students:", len(df))

display(df)
```

```python
# Save dataset as CSV
df.to_csv(
    "student_performance.csv",
    index=False
)

# Load the CSV file
df = pd.read_csv(
    "student_performance.csv"
)

print("Dataset loaded successfully!")

display(df.head(10))
```

```python
print("DATASET DIMENSIONS")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()
```

```python
print("NUMERICAL DATASET SUMMARY")

display(
    df.describe()
)
```

```python
numerical_columns = df.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object"]
).columns.tolist()

print("NUMERICAL COLUMNS:")
print(numerical_columns)

print("\nCATEGORICAL COLUMNS:")
print(categorical_columns)
```

```python
# Select input feature columns

feature_columns = [
    "study_hours",
    "attendance_pct",
    "assignment_score",
    "previous_exam_score"
]

X = df[feature_columns].copy()

print("SELECTED INPUT FEATURES (X)")

display(X.head(10))

print("\nFeature column names:")
print(X.columns.tolist())

print("\nFeature dataset shape:")
print(X.shape)
```

```python
# Select target column

target_column = "final_exam_score"

y = df[target_column].copy()

print("SELECTED TARGET VARIABLE (y)")

display(y.head(10))

print("\nTarget column name:")
print(target_column)

print("\nTarget dataset shape:")
print(y.shape)
```

```python
feature_explanations = pd.DataFrame({
    "Feature": [
        "study_hours",
        "attendance_pct",
        "assignment_score",
        "previous_exam_score"
    ],

    "Role": [
        "Input Feature",
        "Input Feature",
        "Input Feature",
        "Input Feature"
    ],

    "Reason for Selection": [
        "Represents the time spent studying and may help explain academic performance.",

        "Represents class attendance and may provide information about student participation.",

        "Represents assignment performance and demonstrates academic understanding.",

        "Represents previous academic performance and may help predict future examination results."
    ]
})

print("FEATURE SELECTION REASONS")

display(feature_explanations)
```

```python
print("=" * 55)
print("       FEATURES AND TARGET VALIDATION")
print("=" * 55)

# Verify target is not included in X
if target_column not in X.columns:
    print("SUCCESS: Target is not included in X.")

# Verify number of records
if len(X) == len(y):
    print("SUCCESS: Feature and target records match.")

# Verify missing values
print("\nMissing values in X:")
print(X.isnull().sum())

print("\nMissing values in y:")
print(y.isnull().sum())

# Display final shapes
print("\nX shape:", X.shape)
print("y shape:", y.shape)

print("\nSelected Features:")
print(X.columns.tolist())

print("\nSelected Target:")
print(target_column)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("=" * 55)
print("           TRAIN-TEST SPLIT RESULTS")
print("=" * 55)

print("\nTraining feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)

print("\nTraining target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

print("\nTraining input features:")
display(X_train.head())

print("\nTesting input features:")
display(X_test.head())

print("\nTraining target values:")
display(y_train.head())

print("\nTesting target values:")
display(y_test.head())
```

```python
selection_summary = pd.DataFrame({
    "Component": [
        "Input Feature 1",
        "Input Feature 2",
        "Input Feature 3",
        "Input Feature 4",
        "Target Variable"
    ],

    "Selected Column": [
        "study_hours",
        "attendance_pct",
        "assignment_score",
        "previous_exam_score",
        "final_exam_score"
    ],

    "Role": [
        "Input (X)",
        "Input (X)",
        "Input (X)",
        "Input (X)",
        "Output (y)"
    ],

    "Explanation": [
        "Represents time spent studying.",
        "Represents student attendance.",
        "Represents assignment performance.",
        "Represents previous academic performance.",
        "Final examination score to be predicted."
    ]
})

print("FINAL FEATURE AND TARGET SELECTION SUMMARY")

display(selection_summary)
```

```python
# Keep student ID for record identification.
# Use only the four selected numerical columns as X.

selected_dataset = df[
    ["student_id"] +
    feature_columns +
    [target_column]
].copy()

print("SELECTED MACHINE LEARNING DATASET")

display(selected_dataset)
```

```python
# Create project output directory if needed
Path("screenshots").mkdir(exist_ok=True)

# Original dataset
df.to_csv(
    "student_performance.csv",
    index=False
)

# Selected features
X.to_csv(
    "X_features.csv",
    index=False
)

# Selected target with student IDs
y_export = pd.DataFrame({
    "student_id": df["student_id"],
    "final_exam_score": y
})

y_export.to_csv(
    "y_target.csv",
    index=False
)

# Combined selected dataset
selected_dataset.to_csv(
    "selected_dataset.csv",
    index=False
)

# Feature selection explanation
feature_explanations.to_csv(
    "feature_selection_summary.csv",
    index=False
)

print("All project files exported successfully!")
```
